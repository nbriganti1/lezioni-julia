**Esercizio 4.** Dato un intero positivo `n`, disegnare sul piano complesso le radici `n` esime dell'unita'. Dentro il comando `scatter` o `plot`, usare `aspect_ratio=:equal` per avere stessa scala sui 2 assi cartesiani. Inoltre i comandi `xlims=(-1.2, 1.2)`, `ylims=(-1.2, 1.2)`, sempre dentro `scatter` o `plot` permettono di dare il range degli assi.

````julia



n = 6
k = 0:(n-1)

# Calcolo delle radici n-esime dell'unità
radici = exp.(2 * π * im .* k ./ n)

# Disegno dei punti
scatter(
    real.(radici),
    imag.(radici),
    color=:blue,
    aspect_ratio=:equal,
    xlims=(-1.2, 1.2),
    ylims=(-1.2, 1.2)
)
````

**Esercizio 5.** Dato un numero complesso `c` e un numero reale positivo `r`, tracciare la circonferenza di centro `c` e raggio `r`. Dentro il `plot`, scrivere `aspect_ratio=:equal` per avere stessa scala sui 2 assi cartesiani.


````julia

using Plots
c = 2 + 3im
r = 1.5
θ = range(0, 2π, length=300)

# 3. Calcolo dei punti della circonferenza: z(θ) = c + r * e^(i*θ)
z = c .+ r .* exp.(im .* θ)

# 4. Tracciamento del grafico
plot(real.(z),
    imag.(z),
    label="Circonferenza",
    color=:blue,
    linewidth=2,
    aspect_ratio=:equal,
    xlabel="Parte Reale",
    ylabel="Parte Immaginaria",)
# Evidenziamo il centro c
scatter!([real(c)], [imag(c)], color=:red, marker=:circle, label="Centro c")
````

# 3. Traslazioni e rotazioni sul piano

Queste trasformazioni si effettuano con l'aritmetica complessa. Dato il vettore rappresentato dal numero complezzo $z$:

*   Traslazione di $w$: $z\rightarrow z + w$
*   Rotazione di angolo $\theta$: $z\rightarrow z \cdot e^{i\theta}$

**Esempio.** Per ruotare di un angolo $\pi/4$ il quadrato con vertici definiti dal vettore
`quadrato = [0+0im, 1+0im, 1+1im, 0+1im]` e traslarlo di un vettore `v = 3+2i` attorno al baricentro si possono dare i seguenti comandi

```julia
quadrato = [0+0im, 1+0im, 1+1im, 0+1im]
v = 3 +2im
t = pi/4
ruotato = quadrato.* exp(im * t) # Applica rotazione di π/4
traslato_ruotato = ruotato .+ v  # Applica traslazione

# aggiunge il primo elemento per chiudere il poligono
quadrato = push!(quadrato, quadrato[1])
traslato_ruotato = push!(traslato_ruotato, traslato_ruotato[1])

# Disegna il quadrato originale e quello trasformato
plot(real.(quadrato), imag.(quadrato), label="Originale", aspect_ratio=:equal)
plot!(real.(traslato_ruotato), imag.(traslato_ruotato), label="Trasformato", color=:red)

````julia
quadrato = [0+0im, 1+0im, 1+1im, 0+1im]
v = 3 +2im
t = pi/4
ruotato = quadrato.* exp(im * t) # Applica rotazione di π/4
traslato_ruotato = ruotato .+ v  # Applica traslazione

# aggiunge il primo elemento per chiudere il poligono
quadrato = push!(quadrato, quadrato[1])
traslato_ruotato = push!(traslato_ruotato, traslato_ruotato[1])

# Disegna il quadrato originale e quello trasformato
plot(real.(quadrato), imag.(quadrato), label="Originale", aspect_ratio=:equal)
plot!(real.(traslato_ruotato), imag.(traslato_ruotato), label="Trasformato", color=:red)



````

**Esercizio 7**. Ruotare di un angolo di $\pi/6$, intorno al proprio baricento (calcolato come media dei vertici),  il quadrilatero con vertici definiti dal vettore `z = [0, 1, 1+2im, 3im]`. Per il calcolo della media si usi l'istruzione `sum`.



````julia


using Plots

z = [0, 1, 1+2im, 3im]
zc = sum(z) / length(z)

t = π / 6
z_ruotato = zc .+ (z .- zc) .* exp(im * t)

# 5. Per tracciare i poligoni chiusi, aggiungiamo il primo vertice in coda
z_chiuso = [z..., z[1]]
z_ruotato_chiuso = [z_ruotato..., z_ruotato[1]]

# 6. Grafico
plot(
    real.(z_chiuso),
    imag.(z_chiuso),
    label="Quadrilatero",
    color=:blue,
    linewidth=2,
    aspect_ratio=:equal,
    xlabel="Parte Reale",
    ylabel="Parte Immaginaria",
    title="Rotazione di π/6 attorno al baricentro"
)

plot!(
    real.(z_ruotato_chiuso),
    imag.(z_ruotato_chiuso),
    label="Quadrilatero ruotato",
    color=:red,
    linewidth=2,
    linestyle=:dash
)

# Evidenziamo il baricentro
scatter!([real(zc)], [imag(zc)], color=:black, label="Baricentro")
````

# 4. Trasformazioni di Moebius

Le funzioni di variabile complessa del tipo
$$
z\longrightarrow\frac{az+b}{cz+d},
$$
con $ad-bc\neq 0$, sono note come *trasformazioni di Moebius*. (Rotazioni e traslazioni sono casi particolari di queste trasformazioni).


**Esercizio 8**. Svolgere le seguenti operazioni:


*  definire un vettore `v` di numeri complessi che rappresentino ``tanti'' punti equispaziati sulla circonferenza di centro $1$ e raggio $1$, ad esempio 100.
* disegnare la circonferenza usando il vettore `v`
* applicare a ciascun elemento di `v` la trasformazione di Moebius  
$$
z\longrightarrow\frac{az+b}{z+1},
$$
dove $a$ e $b$ sono rispettivamente la penultima e l'ultima cifra del vostro numero di matricola (se sono uguali prendete invece $a=1$ e $b=0$), ottenendo un nuovo vettore `w`,
* rappresentare i punti di `w` sullo stesso grafico usato in precedenza (Per disegnare sullo stesso grafico dare il comando `plot!`.

Qual e` la figura che ne risulta?




````julia
# Inserire qui il codice
using Plots

# 1. Vettore v di 100 punti equispaziati sulla circonferenza di centro 1 e raggio 1
N = 100
θ = range(0, 2π, length=N)
v = 1 .+ exp.(im .* θ)

# 2. Grafico della circonferenza originale
plot(
    real.(v),
    imag.(v),
    label="Circonferenza originale",
    color=:blue,
    aspect_ratio=:equal,
    xlabel="Parte Reale",
    ylabel="Parte Immaginaria",
    title="Trasformazione di Möbius"
)

# 3. Definiamo i parametri a e b (sostituisci con le cifre della tua matricola)
# Esempio con a = 2 e b = 3 (se uguali imposta a = 1, b = 0):
a = 7
b = 2
w = (a .* v .+ b) ./ (v .+ 1)

# 4. Tracciamento del nuovo vettore w sullo stesso grafico
plot!(
    real.(w),
    imag.(w),
    label="Trasformazione w",
    color=:red,
    linestyle=:dash)
````
